#### 1. Create an array `x` from 0 to 10 and compute `y = x**2`

```python
import numpy as np
```

```python
x = np.arange(0,10+1)
print(x)
```

```python
y = np.array(x**2)
print(y)
```

#### 2. Generate 100 random values

```python
random_val = np.random.randint(0,100,100)
print(random_val)
```

```python
print(f"Mean: {np.mean(random_val)}")
print(f"Standard Deviation: {np.std(random_val)}")
```
